```python

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def softmax(a):
    c = np.max(a)
    exp_a = np.exp(a - c)
    sum_exp_a = np.sum(exp_a)
    y = exp_a / sum_exp_a
    return y

# 입력값
X = np.array([1, 2])

# 3번: 1층
W1 = np.array([[2, 1], [-1, 1]])
B1 = np.array([1, 1])

A1 = np.dot(X, W1) + B1
Z1 = sigmoid(A1)

print("3번 A1:", A1)
print("3번 Z1:", Z1)

# 4번: 2층
W2 = np.array([[0, 1], [2, 1]])
B2 = np.array([1, -1])

A2 = np.dot(Z1, W2) + B2
Z2 = sigmoid(A2)

print("4번 A2:", A2)
print("4번 Z2:", Z2)

# 5번: 출력층
W3 = np.array([[1, -2], [2, 1]])
B3 = np.array([-1, 1])

A3 = np.dot(Z2, W3) + B3
Y = softmax(A3)

print("5번 A3:", A3)
print("5번 Y:", Y)
```

```python

import sys, os
sys.path.append(os.getcwd())

import numpy as np
import pickle

from dataset.mnist import load_mnist
from common.functions import sigmoid, softmax


# MNIST 테스트 데이터 불러오기
def get_data():
    (x_train, t_train), (x_test, t_test) = \
        load_mnist(normalize=True, flatten=True, one_hot_label=False)

    return x_test, t_test


# 학습된 가중치와 편향 불러오기
def init_network():
    with open("sample_weight.pkl", "rb") as f:
        network = pickle.load(f)

    return network


# 신경망 순전파
def predict(network, x):
    W1, W2, W3 = network['W1'], network['W2'], network['W3']
    b1, b2, b3 = network['b1'], network['b2'], network['b3']

    a1 = np.dot(x, W1) + b1
    z1 = sigmoid(a1)

    a2 = np.dot(z1, W2) + b2
    z2 = sigmoid(a2)

    a3 = np.dot(z2, W3) + b3
    y = softmax(a3)

    return y


# 데이터와 신경망 불러오기
x, t = get_data()
network = init_network()


# 정확도 계산
accuracy_cnt = 0

for i in range(len(x)):
    y = predict(network, x[i])
    p = np.argmax(y)

    if p == t[i]:
        accuracy_cnt += 1


# 결과 출력
print("Accuracy:" + str(float(accuracy_cnt) / len(x)))
```
